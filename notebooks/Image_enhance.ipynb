```python
import os
import cv2
```

```python
os.makedirs("/content/drive/MyDrive/image-authentication-project/data/edited", exist_ok=True)
```

```python
actual_folder = "/content/drive/MyDrive/image-authentication-project/data/actual/actual_images"
edited_folder = "/content/drive/MyDrive/image-authentication-project/data/edited"

for file in os.listdir(actual_folder):

    img_path = os.path.join(actual_folder, file)

    img = cv2.imread(img_path)

    # skip non-image files
    if img is None:
        continue

    # 1. resize
    img = cv2.resize(img, (256, 256))

    # 2. blur (manipulation)
    img = cv2.GaussianBlur(img, (5, 5), 0)

    # 3. brightness change
    img = cv2.convertScaleAbs(img, alpha=1.3, beta=20)

    # save output
    save_path = os.path.join(edited_folder, file)
    cv2.imwrite(save_path, img)

    print("Saved:", save_path)
```
