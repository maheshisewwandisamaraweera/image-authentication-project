```python
import torch
from diffusers import StableDiffusionImg2ImgPipeline
```

```python
pipe = StableDiffusionImg2ImgPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)
```

```python
pipe = pipe.to("cuda")
```

```python
print("Model loaded successfully:", pipe)
```

```python
import os

actual_folder = "/content/drive/MyDrive/image-authentication-project/data/actual/actual_images"
ai_folder = "/content/drive/MyDrive/image-authentication-project/data/ai_generated"

os.makedirs(ai_folder, exist_ok=True)
```

```python
from PIL import Image

prompt = "highly realistic AI generated image, detailed textures, synthetic style, enhanced lighting"

for file in os.listdir(actual_folder):

    img_path = os.path.join(actual_folder, file)

    try:
        init_image = Image.open(img_path).convert("RGB").resize((512, 512))
    except:
        continue

    # AI generation
    result = pipe(
        prompt=prompt,
        image=init_image,
        strength=0.75,
        guidance_scale=7.5
    ).images[0]

    save_path = os.path.join(ai_folder, file)
    result.save(save_path)

    print("Saved:", save_path)
```
