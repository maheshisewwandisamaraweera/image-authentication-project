```python
import os
import torch
from PIL import Image
from diffusers import StableDiffusionImg2ImgPipeline
```

```python
pipe = StableDiffusionImg2ImgPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")
```

```python
input_folder = "/content/drive/MyDrive/image-authentication-project/data/actual"

output_folder = "/content/drive/MyDrive/image-authentication-project/data/ai_generated"

os.makedirs(output_folder, exist_ok=True)
```

```python
prompt = "A realistic high quality photograph, natural lighting, detailed image"

for filename in os.listdir(input_folder):

    if filename.lower().endswith((".jpg",".jpeg",".png")):

        print("Processing:", filename)

        image_path = os.path.join(input_folder, filename)

        image = Image.open(image_path).convert("RGB")

        # Required size for Stable Diffusion
        image = image.resize((512,512))


        generated_image = pipe(
            prompt=prompt,
            image=image,
            strength=0.35,
            guidance_scale=7.5
        ).images[0]


        save_path = os.path.join(
            output_folder,
            "AI_" + filename
        )

        generated_image.save(save_path)


print("AI-generated images completed!")
```
